```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
import pandas as pd
```

```python
texts = [
    # Positive
    "The movie was excellent and entertaining",
    "I loved the amazing movie",
    "The film was wonderful and enjoyable",
    "The movie was fantastic",
    "I really enjoyed watching this film",
    "The movie was exciting and interesting",
    "This was a great movie",
    "The film was impressive and enjoyable",
    "I had a wonderful movie experience",
    "The movie was brilliant",
    "The food was delicious and fresh",
    "The restaurant served excellent food",
    "I loved the tasty meal",
    "The food was wonderful",
    "The meal was fresh and delicious",
    "The restaurant food was amazing",
    "I really enjoyed the delicious food",
    "The meal was excellent",
    "The food tasted great",
    "The restaurant was fantastic",
    "This product works perfectly",
    "I am very satisfied with this product",
    "The product quality is excellent",
    "I love this product",
    "This product is reliable and useful",
    "The product has excellent quality",
    "I am happy with my purchase",
    "This product is amazing",
    "The product performs very well",
    "I would recommend this product",
    "The service was excellent",
    "The staff were friendly and helpful",
    "I am very happy with the service",
    "The customer support was wonderful",
    "The service was fast and efficient",
    "The staff provided excellent support",
    "I received great customer service",
    "The support team was helpful",
    "The service was professional",
    "I am satisfied with the customer support",

    # Negative
    "The movie was terrible and boring",
    "I hated the awful movie",
    "The film was disappointing",
    "The movie was horrible",
    "I did not enjoy watching this film",
    "The movie was dull and boring",
    "This was a terrible movie",
    "The film was disappointing and slow",
    "I had a horrible movie experience",
    "The movie was awful",
    "The food was terrible and cold",
    "The restaurant served horrible food",
    "I hated the bad meal",
    "The food was disgusting",
    "The meal was cold and tasteless",
    "The restaurant food was awful",
    "I did not enjoy the terrible food",
    "The meal was disappointing",
    "The food tasted horrible",
    "The restaurant was terrible",
    "This product does not work",
    "I am very disappointed with this product",
    "The product quality is terrible",
    "I hate this product",
    "This product is unreliable and useless",
    "The product has poor quality",
    "I am unhappy with my purchase",
    "This product is horrible",
    "The product performs very badly",
    "I would not recommend this product",
    "The service was terrible",
    "The staff were rude and unhelpful",
    "I am very unhappy with the service",
    "The customer support was horrible",
    "The service was slow and inefficient",
    "The staff provided poor support",
    "I received terrible customer service",
    "The support team was unhelpful",
    "The service was unprofessional",
    "I am dissatisfied with the customer support"
]

labels = ["Positive"] * 40 + ["Negative"] * 40

print("Total texts:", len(texts))
print("Total labels:", len(labels))

print("\nClass Distribution:")
print(pd.Series(labels).value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    texts,
    labels,
    test_size=0.20,
    random_state=42,
    stratify=labels
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2)
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("TF-IDF Training Shape:", X_train_tfidf.shape)
print("TF-IDF Testing Shape:", X_test_tfidf.shape)
```

```python
print("Number of TF-IDF Features:",
      len(vectorizer.get_feature_names_out()))

print("\nFirst 50 Features:")
print(vectorizer.get_feature_names_out()[:50])
```

```python
model = SVC(
    kernel="linear",
    C=1.0
)

model.fit(X_train_tfidf, y_train)

print("SVM model trained successfully.")
```

```python
y_pred = model.predict(X_test_tfidf)

print("Actual Labels:")
print(y_test)

print("\nPredicted Labels:")
print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
print(
    "\nClassification Report:\n",
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
results = pd.DataFrame({
    "Text": X_test,
    "Actual": y_test,
    "Predicted": y_pred
})

print(results.to_string(index=False))
```

```python
new_texts = [
    "The movie was fantastic and enjoyable",
    "The food was horrible and disgusting",
    "This product works perfectly",
    "The product is useless and terrible",
    "The customer service was excellent",
    "The customer service was awful"
]

new_texts_tfidf = vectorizer.transform(new_texts)

new_predictions = model.predict(new_texts_tfidf)

print("New Text Predictions:\n")

for text, prediction in zip(new_texts, new_predictions):
    print("Text:", text)
    print("Prediction:", prediction)
    print()
```
